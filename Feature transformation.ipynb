# What is Nominal Data?

# How to Explain Nominal Data in an Interview?

### Nominal data is a type of categorical data used to label or classify data points without any order or rank.
### In simple terms, it’s just names or labels that represent different categories.
## Example:
### Imagine we collect data about people’s favorite fruits.
### The options could be: Apple, Mango, Banana, Orange.
### These are just labels with no specific order—no fruit is "greater" than another.

# Key Characteristics of Nominal Data:
### No Numeric Meaning: You can't perform arithmetic (like adding or averaging) on nominal data.
### No Order or Rank: The categories don’t have a meaningful order.
### Categories must be distinct: Each value must fit into only one category.


```python
#Examples of Nominal Data:
#Colors: Red, Blue, Green
#Gender: Male, Female, Non-Binary
#Cities: Delhi, Mumbai, Bangalore
#Marital Status: Single, Married, Divorced
```

# How to Use Nominal Data in Analysis?

### Encoding: In machine learning, you can convert nominal data into numbers using techniques like one-hot encoding.
### Visualization: Nominal data is often visualized using bar charts or pie charts.


# Converting Nominal Data to One-Hot Encoding:

```python
import pandas as pd

# Create a sample dataset with nominal data
data = {'City': ['Mumbai', 'Delhi', 'Bangalore', 'Mumbai']}
df = pd.DataFrame(data)

# Convert the 'City' column to one-hot encoded columns and ensure integers
one_hot = pd.get_dummies(df['City']).astype(int)

print(one_hot)

```

# What is Ordinal Data?
### Ordinal data is a type of categorical data where the categories have a meaningful order or rank, but the differences between the categories are not measurable.
### In simple terms, ordinal data shows position or rank but doesn’t tell how much difference exists between them.

# Examples of Ordinal Data:
### Rating levels: Poor, Average, Good, Excellent
### Education levels: High School, Bachelor’s, Master’s, PhD
### Customer satisfaction: Very Unsatisfied, Unsatisfied, Neutral, Satisfied, Very Satisfied
### Size: Small, Medium, Large, Extra Large


# How to Use Ordinal Data in Data Science?
### Encoding: Convert ordinal data into numbers to use it in models (e.g., Small = 1, Medium = 2, Large = 3).
### Visualization: Use bar charts or histograms to analyze ordinal data.

# Ordinal Encoding Example: Movie Ratings
### Imagine we have data where users rate movies with the following levels:
### Bad, Average, Good, Excellent.

### Since these ratings have a natural order (Bad < Average < Good < Excellent), we can use ordinal encoding.

```python
import pandas as pd

# Create a sample dataset with ordinal data
data = {'Movie_Rating': ['Good', 'Bad', 'Excellent', 'Average', 'Good']}
df = pd.DataFrame(data)

# Define a custom mapping for ordinal encoding
rating_mapping = {'Bad': 1, 'Average': 2, 'Good': 3, 'Excellent': 4}

# Apply the mapping to the 'Movie_Rating' column
df['Rating_Encoded'] = df['Movie_Rating'].map(rating_mapping)

print(df)
```

# What is Encoding in Data Science?
### Encoding is the process of converting categorical data (like words or labels) into numbers so that a machine learning model can understand it.

### Why? Computers and ML models can only work with numbers, not text or categories.

# Types of Encoding with Simple Examples
### 1. One-Hot Encoding
#### Each unique category gets a new binary column (1 or 0).
#### Use when categories have no order (like colors or cities).

#### Example: Imagine we have a column with colors: Red, Green, Blue.

```python
import pandas as pd

data = {'Color': ['Red', 'Green', 'Blue', 'Red']}
df = pd.DataFrame(data)

# One-hot encoding
encoded_df = pd.get_dummies(df['Color']).astype(int)

print(encoded_df)
```

```python
#Here, Red becomes [0, 0, 1], Green becomes [0, 1, 0], and so on.
```

## 2. Ordinal Encoding
#### Each category is converted into numbers based on a specific order or rank.
#### Use when categories have an order (like ratings or sizes).
#### Example: Consider a dataset with sizes: Small, Medium, Large.

```python
size_mapping = {'Small': 1, 'Medium': 2, 'Large': 3}
sizes = ['Medium', 'Small', 'Large', 'Small']

# Applying ordinal encoding
encoded_sizes = [size_mapping[size] for size in sizes]

print(encoded_sizes)
```

## Why Encoding is Important?
#### ML models need numbers to process the data.
#### Encoding makes text-based or categorical data usable in models like regression, decision trees, etc.
#### Different types of encoding are used depending on the type of data (ordered or unordered).
