```python
###Graphs for Continuous Data

#Line Graph
#Definition: Shows data points connected by lines.
#Use: Good for tracking changes over time (e.g., temperature over a week).

#Scatter Plot
#Definition: Displays individual data points as dots on a two-dimensional graph.
#Use: Useful for showing relationships between two continuous variables (e.g., height vs. weight).

#Histogram
#Definition: Groups continuous data into bins and shows how many data points fall into each bin.
#Use: Great for showing the distribution of data (e.g., age distribution in a population).

#Box Plot (Box-and-Whisker Plot)
#Definition: Displays the median, quartiles, and outliers of a dataset.
#Use: Helpful for comparing distributions across different groups (e.g., test scores by class).

#Density Plot
#Definition: A smoothed version of a histogram that shows the distribution of data.
#Use: Good for visualizing the distribution without the bins (e.g., income distribution).

#Area Chart
#Definition: Similar to a line graph but filled with color below the line.
#Use: Useful for showing cumulative totals over time (e.g., total sales growth).

#Heatmap
#Definition: Uses color to show the density or intensity of data points in two dimensions.
#Use: Great for visualizing correlations or patterns in large datasets (e.g., correlation matrix).

###Graphs for Discrete Data

#Bar Chart
#Definition: Displays bars representing different categories with heights corresponding to their values.
#Use: Ideal for comparing quantities across different categories (e.g., number of students in each grade).

#Pie Chart
#Definition: A circular chart divided into slices to show proportions of a whole.
#Use: Useful for showing parts of a whole (e.g., survey responses).

#Dot Plot
#Definition: Uses dots to represent individual data points along a number line.
#Use: Good for displaying small discrete datasets (e.g., results of dice rolls).

#Stacked Bar Chart
#Definition: Shows total values as bars divided into segments representing different categories.
#Use: Useful for showing composition within categories (e.g., sales by product type in each quarter).

#Grouped Bar Chart
#Definition: Displays bars for multiple groups side by side for comparison.
#Use: Helpful for comparing categories across different groups (e.g., test scores by gender).

#Frequency Polygon
#Definition: A line graph that connects points representing the frequency of each category.
#Use: Useful for comparing distributions of discrete data (e.g., number of customers per age group).
```

## correlation

```python
#Correlation studies and measures the direction and extent of relationship among variables 
#correlation summarize the relationship between two variables in a single number called the correlation coefficient.
#The correlation coefficient is usually represented using the symbol r, and it ranges from -1 to +1.

#A correlation coefficient quite close to 0, but either positive or negative, implies little or no relationship between the two variables.
#A correlation coefficient close to plus 1 means a positive relationship between the two variables, with increases in one of the variables 
#being associated with increases in the other variable.

#A correlation coefficient close to -1 indicates a negative relationship between two variables, 
#with an increase in one of the variables being associated with a decrease in the other variable.
```

## Heat map

```python
#Heatmap is defined as a graphical representation of data using colors to visualize the value of the matrix.
#in this to represent more common values or higher activities brighter
#colors basically less common or activity values,darker colors are preferred

#correlation between the different parameters
data= df.drop('species',axis = 1) #dropping species column since its catagorial column
tc = data.corr()  #find correlation only for continuous values
tc
```

```python
sns.heatmap(tc,annot = True)  #heat map will show color ways representation... in this observation light color says positive relationship and darker says negetive 
```
