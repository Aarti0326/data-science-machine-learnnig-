### EDA in Univarite analysis (analysis a single column at a time)

### import library

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('train (1).csv')
```

```python
df
```

```python
df.head()
```

### 1) catagorical data
#a) countplot

```python
sns.countplot(x = 'Survived',data = df)	
df['Survived'].value_counts()
```

```python
sns.countplot(x = 'Pclass',data = df)
df['Pclass'].value_counts()
```

```python
sns.countplot(x = 'Embarked',data=df)
df['Embarked'].value_counts()
```

```python
df.head()
```

```python
sns.countplot(x = 'Sex',data=df)
df["Sex"].value_counts()
```

### b)piechart

```python
df['Sex'].value_counts().plot(kind='pie',autopct='%.1f')
```

### Numerical data

### a) Histogram

```python
plt.hist(df['Age'],bins = 8)
```

### b) Distplot/Histplot

```python
sns.distplot(df['Age'])
```

```python
sns.histplot(df['Age'])
```

### c) Boxplot

```python
sns.boxplot(df['Age'])
```
