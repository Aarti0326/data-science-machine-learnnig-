## Grouping the data

```python
data.groupby(data['Pclass']).count()  #non null values it will count
```

```python
data.groupby(data['Pclass']).max('Fare') #max function will return the raw which has given maximum Fare
```

```python
data.groupby(data['Pclass']).min('Fare')  #Fare means ticket price
```

```python
data.groupby(data['Pclass']).agg({'Fare':'max','Age':'max'})
```

```python
data.groupby(data['Pclass']).agg({'Fare':'count','Age':'count'})
```
